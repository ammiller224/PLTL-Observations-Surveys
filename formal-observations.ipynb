### Create csv for PLTL subjects from formal observation responses

```python
import csv
import os

parent_dir = os.getcwd()
print(f"Parent Directory: {parent_dir}")
data_dir = os.path.join(parent_dir, 'data')
csv_files = [os.path.join(data_dir, f) for f in os.listdir(data_dir) if f.lower().endswith('.csv')]

subjects = set()
subject_files = {}
selected_range = (17, 53)
excluded_columns = {'Observation Date (mm/dd/yyyy)', 'Time of Session/Observation (##:## AM/PM)'}
for csv_filename in csv_files:
    print(f"Processing file: {csv_filename}")
    with open(csv_filename, newline='', encoding='latin1') as formalcsv:
        formalobs = csv.DictReader(formalcsv)
        headers = list(formalobs.fieldnames)
        selected_columns = [
            col for col in headers[selected_range[0]:selected_range[1]] if col not in excluded_columns
        ]  # Filter out excluded columns
        next(formalobs)
        for row in formalobs:
            date_value = row.get('Observation Date (mm/dd/yyyy)', '').strip()
            
            if row['Session Subject']:
                subject = row['Session Subject']
                subjects.add(subject)

                column_names = list(row.keys())

                if subject not in subject_files:
                    #with open(row['Q4'] + '.csv', 'w') as file:
                    csv_name = os.path.join(parent_dir, subject + '.csv')
                    txt_name = os.path.join(parent_dir, f"{subject}.txt")

                    # open in append mode so files persist across CSVs
                    csv_file = open(csv_name, 'a', newline='', encoding='utf-8')
                    txt_file = open(txt_name, 'a', encoding='utf-8')
                    subject_files[subject] = {'csv': csv_file, 'txt': txt_file}

                    if os.path.getsize(csv_name) == 0:
                        writer = csv.writer(csv_file)
                        writer.writerow(headers)
                writer = csv.writer(subject_files[subject]['csv'])
                #print(row['Session Subject'], subject_files[subject], subject)

                # Write the row data to the corresponding subject file
                #writer = csv.writer(subject_files[subject]['csv'])
                #writer.writerow(row.values())  # Write the row content
                row_values = [row.get(col, '') if row.get(col) != '' else '' for col in row.keys()]
                writer.writerow(row_values)  # Write row content

                # Select row data for text file (skip empty fields)
                selected_row_data = [row.get(col, '') for col in selected_columns]  # Get all data from selected columns
                # Filter out empty fields for the txt file (ensure proper alignment)
                selected_row_data = [(col_name, col_value) for col_name, col_value in zip(selected_columns, selected_row_data) if col_value.strip()]
                txt_writer = subject_files[subject]['txt']

                for col_name, col_value in selected_row_data:
                    txt_writer.write(f"{col_name} : {col_value}\n")
                txt_writer.write("\n")

for files in subject_files.values():
    files['csv'].close()
    files['txt'].close()
```
